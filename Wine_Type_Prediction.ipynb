```python
import pandas as pd
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

Step 2 – Prepare features and target

```python
target = df['type']
X = df.drop('type', axis=1)
```

```python
target.value_counts()
```

```python
# Map target values safely from df['type'] to ensure clean categories
target_mapping = {'red': 0, 'white': 1}
target = df['type'].map(target_mapping)

target.value_counts()
```

Step 3 – Split and scale

```python
# Split dataset
X_train, X_test, y_train, y_test = train_test_split(X, target, test_size=0.2, random_state=42, stratify=target)
```

```python
# Standard Scaler
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

Step 4 – Train and compare kernels

```python
def train_and_evaluate_svm(kernel, degree=3, C=1.0, show_report=True):
    model = SVC(C=C, kernel=kernel, degree=degree, random_state=42)
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    accuracy = accuracy_score(y_test, y_pred)
    print(f"Kernel: {kernel} (Degree: {degree}, C: {C})")
    print("Accuracy:", round(accuracy, 3))
    print('\n')

    if show_report:
        print("\nClassification Report:\n", classification_report(y_test, y_pred))
        print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

```python
train_and_evaluate_svm('linear', 0, 1.0)
```

```python
train_and_evaluate_svm('rbf', 0, 1.0)
```

```python
train_and_evaluate_svm(kernel='poly', degree=3)
```

```python
C = [0.5, 5]

for c in C:
    print(f"C: {c}\n")
    train_and_evaluate_svm(kernel='linear', degree=0, C=c, show_report=False)
    train_and_evaluate_svm(kernel='poly', degree=3, C=c, show_report=False)
    train_and_evaluate_svm(kernel='rbf', degree=0, C=c, show_report=False)
```
